````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 2 格：建立資料庫與初始資料
# ============================================================

DB_NAME = "coffee_shop.db"

# 連接 SQLite 資料庫
conn = sqlite3.connect(DB_NAME)

try:
    cursor = conn.cursor()

    # 刪除舊資料表，確保每次執行都從初始資料開始
    cursor.execute("DROP TABLE IF EXISTS 訂單")
    cursor.execute("DROP TABLE IF EXISTS 菜單")

    # 建立菜單資料表
    cursor.execute("""
        CREATE TABLE 菜單 (
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            price INTEGER NOT NULL,
            stock INTEGER NOT NULL
        )
    """)

    # 建立訂單資料表
    cursor.execute("""
        CREATE TABLE 訂單 (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            item_id INTEGER NOT NULL,
            quantity INTEGER NOT NULL,
            total_price INTEGER NOT NULL,
            status TEXT NOT NULL,
            created_at TEXT DEFAULT CURRENT_TIMESTAMP,
            FOREIGN KEY (item_id) REFERENCES 菜單(id)
        )
    """)

    # 加入老師指定的三種飲品
    menu_data = [
        (1, "美式咖啡", 50, 5),
        (2, "拿鐵", 65, 4),
        (3, "抹茶歐蕾", 70, 3)
    ]

    cursor.executemany(
        "INSERT INTO 菜單 (id, name, price, stock) VALUES (?, ?, ?, ?)",
        menu_data
    )

    conn.commit()
    print("✅ 資料庫建立成功！")
    print("✅ 已建立菜單、訂單兩張資料表")
    print("✅ 已加入 3 種飲品")

finally:
    conn.close()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 4 格：AI 解讀函式
# 功能：將使用者的話轉成 (動作, 品項, 數量, 原始回覆)
# ============================================================

# 允許的動作與品項
ACTIONS = ["訂購", "取消", "查詢", "統計", "未知"]
ITEMS = ["美式咖啡", "拿鐵", "抹茶歐蕾", "全部", "未知"]

# 暫存已經解讀過的句子，避免重複呼叫 AI
ai_cache = {}

def ai_parse(user_text):
    user_text = user_text.strip()

    if user_text in ai_cache:
        return ai_cache[user_text]

    if not user_text:
        return ("未知", "未知", 1, "輸入為空")

    prompt = """
你是校園咖啡廳的訂單解讀助手。
請把使用者的話轉成：動作,品項,數量

動作只能選：訂購、取消、查詢、統計、未知
品項只能選：美式咖啡、拿鐵、抹茶歐蕾、全部、未知

規則：
1. 訂購包含「訂、買、預約、我要」等意思。
2. 取消訂單時輸出「取消」。
3. 詢問剩餘杯數或庫存時輸出「查詢」。
4. 詢問今天銷售情況時輸出「統計,全部,1」。
5. 沒有說數量時預設為 1。
6. 中文數字要轉成阿拉伯數字。
7. 不認識的飲品或無關的句子輸出「未知,未知,1」。
8. 只輸出三個欄位，不要解釋。
"""

    raw = ""

    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "system", "content": prompt},
                    {"role": "user", "content": user_text}
                ]
            )

            raw = response.choices[0].message.content or ""
            time.sleep(2)
            break

        except Exception as e:
            if "429" in str(e) and attempt < 2:
                print("API 使用量限制，30 秒後重試...")
                time.sleep(30)
            else:
                raw = "API 錯誤：" + str(e)
                break

    # 不直接相信 AI，從回覆中驗證動作與品項
    parts = [x.strip() for x in raw.split(",")]

    action = "未知"
    item = "未知"
    quantity = 1

    if len(parts) >= 2:
        if parts[0] in ACTIONS:
            action = parts[0]

        if parts[1] in ITEMS:
            item = parts[1]

    # 使用正規表示式取得數量
    if len(parts) >= 3:
        match = re.search(r"-?\d+", parts[2])
        if match:
            quantity = int(match.group())

    # 統計一律使用「全部」
    if action == "統計":
        item = "全部"

    # 無法辨識品項，視為未知
    if item == "未知" and action != "統計":
        action = "未知"

    result = (action, item, quantity, raw)
    ai_cache[user_text] = result

    return result

print("✅ AI 解讀函式建立成功")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 5 格：AI 解讀測試
# 測試 8 句，與規格書 7-1 一致
# ============================================================

# 格式：(使用者說的話, 預期動作, 預期品項, 預期數量)
test_cases = [
    ("我要一杯美式咖啡", "訂購", "美式咖啡", 1),
    ("幫我訂兩杯拿鐵", "訂購", "拿鐵", 2),
    ("我要取消拿鐵", "取消", "拿鐵", 1),
    ("抹茶歐蕾還剩幾杯", "查詢", "抹茶歐蕾", 1),
    ("今天各品項賣得怎樣", "統計", "全部", 1),
    ("我要訂四杯拿鐵", "訂購", "拿鐵", 4),
    ("我要訂珍珠奶茶", "未知", "未知", 1),
    ("今天天氣如何", "未知", "未知", 1)
]

passed = 0

for i, (user_text, expected_action, expected_item, expected_qty) in enumerate(test_cases, 1):

    action, item, quantity, raw = ai_parse(user_text)

    # 檢查三個欄位是否符合預期
    is_passed = (
        action == expected_action
        and item == expected_item
        and quantity == expected_qty
    )

    if is_passed:
        passed += 1

    print(f"===== 測試 {i} =====")
    print("使用者輸入：", user_text)
    print("AI 原始回覆：", raw)
    print("系統解讀：", action, item, quantity)
    print("測試結果：", "✅ 通過" if is_passed else "❌ 未通過")
    print()

print(f"AI 解讀測試結果：{passed}/{len(test_cases)} 句通過")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 6 格：功能函式
# 訂購、取消、查詢庫存、銷售統計
# ============================================================

# 訂購功能
def order_drink(name, item, quantity):
    if not name or not name.strip():
        return "請輸入學生姓名"

    if quantity < 1 or quantity > 3:
        return "每次只能訂購 1～3 杯"

    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 查詢飲品資料
        cursor.execute(
            "SELECT id, price, stock FROM 菜單 WHERE name = ?",
            (item,)
        )
        drink = cursor.fetchone()

        if drink is None:
            return "找不到此飲品，目前提供：美式咖啡、拿鐵、抹茶歐蕾"

        item_id, price, stock = drink

        # 檢查每人今日訂購數量
        cursor.execute("""
            SELECT COALESCE(SUM(quantity), 0)
            FROM 訂單
            WHERE student_name = ?
            AND status = ?
            AND DATE(created_at) = DATE('now', 'localtime')
        """, (name.strip(), "已訂購"))

        ordered = cursor.fetchone()
        current_total = ordered[0] if ordered is not None else 0

        if current_total + quantity > 3:
            return f"訂購失敗：每人每天最多 3 杯，目前已訂 {current_total} 杯"

        # 檢查庫存
        if stock < quantity:
            return f"訂購失敗：{item} 庫存不足，目前剩下 {stock} 杯"

        total_price = price * quantity

        # 新增訂單
        cursor.execute("""
            INSERT INTO 訂單
            (student_name, item_id, quantity, total_price, status, created_at)
            VALUES (?, ?, ?, ?, ?, DATETIME('now', 'localtime'))
        """, (name.strip(), item_id, quantity, total_price, "已訂購"))

        # 扣除庫存
        cursor.execute(
            "UPDATE 菜單 SET stock = stock - ? WHERE id = ?",
            (quantity, item_id)
        )

        conn.commit()

        return (
            f"訂購成功！{name.strip()} 訂購 {item} {quantity} 杯，"
            f"總金額 {total_price} 元，目前剩餘 {stock - quantity} 杯"
        )

    finally:
        conn.close()


# 取消訂單功能
def cancel_order(name, item):
    if not name or not name.strip():
        return "請輸入學生姓名"

    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 查詢該學生最新一筆有效訂單
        cursor.execute("""
            SELECT o.id, o.item_id, o.quantity
            FROM 訂單 o
            JOIN 菜單 m ON o.item_id = m.id
            WHERE o.student_name = ?
              AND m.name = ?
              AND o.status = ?
            ORDER BY o.id DESC
            LIMIT 1
        """, (name.strip(), item, "已訂購"))

        order = cursor.fetchone()

        if order is None:
            return f"找不到 {name.strip()} 可取消的 {item} 訂單"

        order_id, item_id, quantity = order

        # 更新訂單狀態
        cursor.execute(
            "UPDATE 訂單 SET status = ? WHERE id = ?",
            ("已取消", order_id)
        )

        # 恢復庫存
        cursor.execute(
            "UPDATE 菜單 SET stock = stock + ? WHERE id = ?",
            (quantity, item_id)
        )

        conn.commit()

        return f"取消成功！已取消 {item} {quantity} 杯，庫存已恢復"

    finally:
        conn.close()


# 查詢庫存功能
def check_stock(item):
    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        cursor.execute(
            "SELECT stock FROM 菜單 WHERE name = ?",
            (item,)
        )

        result = cursor.fetchone()

        if result is None:
            return "找不到此飲品"

        return f"{item} 目前剩餘 {result[0]} 杯"

    finally:
        conn.close()


# 銷售統計功能
def sales_report():
    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 使用 LEFT JOIN + GROUP BY
        # 確保沒有銷售的品項也會顯示
        cursor.execute("""
            SELECT
                m.name,
                m.price,
                m.stock,
                COALESCE(SUM(
                    CASE WHEN o.status = '已訂購'
                    AND DATE(o.created_at) = DATE('now', 'localtime')
                    THEN o.quantity ELSE 0 END
                ), 0) AS sold,
                COALESCE(SUM(
                    CASE WHEN o.status = '已訂購'
                    AND DATE(o.created_at) = DATE('now', 'localtime')
                    THEN o.total_price ELSE 0 END
                ), 0) AS revenue
            FROM 菜單 m
            LEFT JOIN 訂單 o ON m.id = o.item_id
            GROUP BY m.id, m.name, m.price, m.stock
            ORDER BY m.id
        """)

        rows = cursor.fetchall()

        report = "===== 今日銷售統計 =====\n"

        for name, price, stock, sold, revenue in rows:
            report += (
                f"{name}｜單價 {price} 元｜"
                f"已售 {sold} 杯｜剩餘 {stock} 杯｜"
                f"營收 {revenue} 元\n"
            )

        return report

    finally:
        conn.close()


print("✅ 第 6 格功能函式建立成功")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# ============================================================
# 第 7 格：主流程與流程測試
# ============================================================

# 主流程：根據 AI 解讀結果執行對應功能
def main_process(name, user_text):
    if not user_text or not user_text.strip():
        return "請輸入需求"

    action, item, quantity, raw = ai_parse(user_text)

    if action == "訂購":
        return order_drink(name, item, quantity)

    elif action == "取消":
        return cancel_order(name, item)

    elif action == "查詢":
        return check_stock(item)

    elif action == "統計":
        return sales_report()

    else:
        return "無法辨識需求，目前提供：美式咖啡、拿鐵、抹茶歐蕾"


# 流程測試：與規格書 7-2 一致，剛好 8 步
flow_tests = [
    ("阿明", "我要兩杯拿鐵"),
    ("小美", "我要一杯美式咖啡"),
    ("阿明", "我要兩杯美式咖啡"),
    ("阿明", "我要取消拿鐵"),
    ("阿明", "我要三杯抹茶歐蕾"),
    ("小美", "抹茶歐蕾還剩幾杯"),
    ("小美", "我要兩杯拿鐵"),
    ("店長", "今天各品項賣得怎樣")
]

print("===== 流程測試開始 =====")

for i, (name, user_text) in enumerate(flow_tests, 1):
    print(f"\n===== 第 {i} 步 =====")
    print("姓名：", name)
    print("使用者輸入：", user_text)

    result = main_process(name, user_text)

    print("系統回應：", result)

print("\n===== 流程測試完成 =====")

# 最後顯示資料表
show_tables()
````
